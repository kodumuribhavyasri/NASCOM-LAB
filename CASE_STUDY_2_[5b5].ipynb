24. Find \(dF/dx\)

Given:

$$ F(x)=40+5x-0.2x^2 $$

Differentiate:

$$ \frac{dF}{dx}=0+5-0.4x $$

Therefore,

$$ \boxed{F'(x)=5-0.4x} $$

25. Verify the derivative
$$ F'(x)=5-0.4x $$

Hence, the derivative is correct.

26. Find \(x\) when \(F'(x)=0\)
$$ 5-0.4x=0 $$ $$ 0.4x=5 $$ $$ x=\frac{5}{0.4} $$
x=12.5
	​


```python
# Function for predicted final score
def F(x):
    return 40 + 5*x - 0.2*x**2

# Derivative of F(x)
def dF(x):
    return 5 - 0.4*x

# Display values from 1 to 10 study hours
for x in range(1, 11):
    print("Study Hours:", x)
    print("Final Score:", F(x))
    print("Derivative:", dF(x))
    print()
```
